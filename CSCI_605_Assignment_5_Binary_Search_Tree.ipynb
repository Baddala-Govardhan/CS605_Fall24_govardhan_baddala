```python
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    # insert
    def insert(self, name, phone_number):
        value = (name, phone_number)
        self.root = self._insert(self.root, value)

    def _insert(self, node, value):
        if node is None:
            return BSTNode(value)
        if value < node.value:
            node.left = self._insert(node.left, value)
        else:
            node.right = self._insert(node.right, value)
        return node

    # search
    def search(self, name):
        node = self._search(self.root, name)
        if node:
            return node.value[1]
        else:
            return "Name not found"

    def _search(self, node, name):
        if node is None or node.value[0] == name:
            return node
        if name < node.value[0]:
            return self._search(node.left, name)
        return self._search(node.right, name)

    # delete
    def delete(self, name):
        self.root = self._delete(self.root, name)

    def _delete(self, node, name):
        if not node:
            return None

        if name < node.value[0]:
            node.left = self._delete(node.left, name)
        elif name > node.value[0]:
            node.right = self._delete(node.right, name)
        else:
            if not node.left:
                temp = node.right
                node = None
                return temp
            elif not node.right:
                temp = node.left
                node = None
                return temp

            temp = self._minValueNode(node.right)
            node.value = temp.value
            node.right = self._delete(node.right, temp.value[0])

        return node

    # Helper to find minimum value node
    def _minValueNode(self, node):
        current = node
        while current.left is not None:
            current = current.left
        return current

    # in-order traversal
    def inorder_traversal(self):
        result = []
        def inorder(node):
            if node:
                inorder(node.left)
                result.append(node.value)
                inorder(node.right)
        inorder(self.root)
        return result

# Example
bst = BinarySearchTree()
bst.insert("Govardhan", "555-12345")
bst.insert("Bob", "555-2345")
bst.insert("Charlie", "555-3456")
bst.insert("Alice", "555-1234")

# Searching for phone numbers
print("Phone numbers")
print(bst.search("Govardhan"))
print(bst.search("Alice"))
print(bst.search("David"))

# In-order Traversal
print("In-order traversal")
for name, phone in bst.inorder_traversal():
    print(f"{name} - {phone}")

# Deleting a node
bst.delete("Charlie")
print("After deletion of Charlie:")
for name, phone in bst.inorder_traversal():
    print(f"{name} - {phone}")
```
